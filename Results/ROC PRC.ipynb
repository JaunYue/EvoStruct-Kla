```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc, precision_recall_curve, average_precision_score
import ast

with open('targets.txt', 'r') as f:
    targets = ast.literal_eval(f.read())
with open('probs.txt', 'r') as f:
    probs = ast.literal_eval(f.read())

fpr, tpr, _ = roc_curve(targets, probs)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(6,5), dpi=400)
plt.plot(fpr, tpr, color='#B883D3', lw=2, label=f'EvoStruct-Kla (AUC = {roc_auc:.4f})')
plt.plot([0, 1], [0, 1], color='gray', lw=1, linestyle='--')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic Curve')
plt.legend(loc="lower right")
plt.grid(True)
plt.tight_layout()
plt.show()


precision, recall, _ = precision_recall_curve(targets, probs)
pr_auc = average_precision_score(targets, probs)

plt.figure(figsize=(6,5), dpi=400)
plt.plot(recall, precision, color='#B883D3', lw=2, label=f'EvoStruct-Kla (AUPRC = {pr_auc:.4f})')
plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve')
plt.legend(loc="lower left")
plt.grid(True)
plt.tight_layout()
plt.show()
```
